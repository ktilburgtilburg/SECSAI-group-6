```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/twitter_profiles.csv")
df.head()
```

```python
# Volume

print("Rows:", len(df))
print("Columns:", df.shape[1])
print("Total values:", df.size)
print("Memory (MB): %.2f" % (df.memory_usage(deep=True).sum() / 1e6))

print("\nDescriptive statistics:")
print(df.describe())
```

```python
# Velocity

df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

print("Earliest account:", df["created_at"].min())
print("Latest account:", df["created_at"].max())

print("\nAccounts created per year:")
print(df["created_at"].dt.year.value_counts().sort_index())
```

```python
# Variety

print("Data types:")
print(df.dtypes)

print("\nNumber of columns by data type:")
print(df.dtypes.value_counts())
```

```python
print("\nColumn information:")
print(df.nunique().sort_values(ascending=False))
```

```python
# Veracity

print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```
